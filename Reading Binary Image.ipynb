```python
import cv2

# Load the Color/RGB image
image_path = r"D:\From D\NORKEN INTERNATIONAL\R&D Project\Norken Applets\Images\core 1.jpeg"
image1 = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
image2 = cv2.resize(image1, (500,500))

# Threshold the image to ensure it's binary
ret, binary_image = cv2.threshold(image2, 127,255, cv2.THRESH_BINARY)
cv2.imshow("Core 1", binary_image)
cv2.waitKey(0)
cv2.destroyAllWindows()

```
